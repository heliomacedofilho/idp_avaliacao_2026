## 1. O que se espera do trabalho final

O trabalho final é uma avaliação causal de uma política pública, aplicando as técnicas dos Módulos 01, 03 e 04 da disciplina, entregue em **duas peças complementares**:

1. **Um caderno Jupyter (`.ipynb`) que executa todas as etapas do trabalho**, desde o download dos dados brutos nas fontes originais até as tabelas e os gráficos finais. O caderno deve rodar do início ao fim em um computador limpo, sem depender de arquivos preparados fora dele.
2. **Um relatório em formato de artigo científico**, que apresenta pergunta, dados, método, resultados e limitações, e cujas tabelas e figuras saem do caderno.

As duas peças são avaliadas em conjunto. Um artigo bem escrito sobre um caderno que não reproduz os resultados não é aceito; um caderno completo sem um artigo que o interprete também não. O princípio que organiza tudo é o da **reprodutibilidade**: qualquer leitor, com o caderno e acesso à internet, deve chegar aos mesmos números do artigo.

| Peça | Formato | O que contém | Papel na avaliação |
| --- | --- | --- | --- |
| Caderno | `.ipynb` (Python), executado, com saídas salvas | Download, extração, limpeza, construção da base, descritiva, métodos, robustez, exportação de tabelas e figuras | Demonstra que a análise é correta e reprodutível |
| Artigo | `.docx` ou `.pdf` | Resumo, introdução, dados, método, resultados, discussão, conclusão, referências, declaração de uso de IA | Demonstra que o grupo sabe interpretar e comunicar a evidência |

O conteúdo de referência é o do repositório da disciplina, [heliomacedofilho/idp_avaliacao_2026](https://github.com/heliomacedofilho/idp_avaliacao_2026). O Módulo 01 estrutura a pergunta; os Módulos 03 e 04 fornecem os dois métodos de estimação que o trabalho deve aplicar. O Módulo 02 (experimentos aleatorizados) não é exigido, porque poucas políticas brasileiras dispõem de dados experimentais públicos.

| Módulo | Técnicas vistas em aula | O que o trabalho deve conter |
| --- | --- | --- |
| 01 — Fundamentos: causalidade, resultados potenciais e DAGs | Resultados potenciais; ATE, ATT e ATU; SUTVA; viés de seleção; DAGs e caminho de fundos; bons e maus controles (confundidor, mediador, colisor) | Pergunta causal e estimando explícitos; DAG da relação estudada, com os confundidores que motivam as covariáveis e os mediadores e colisores que não devem ser controlados; discussão de SUTVA |
| 03 — Seleção nos observáveis: pareamento e escore de propensão | Ignorabilidade e suporte comum; subclassificação e pareamento exato; vizinho mais próximo; escore de propensão por regressão logística ou *machine learning* calibrado; pareamento, estratificação e IPW pelo escore; balanceamento (SMD) como critério; estimadores duplamente robustos | Método 1: escore estimado, diagnóstico de suporte comum, ao menos um estimador baseado no escore, tabela de balanceamento antes e depois, estimativa com intervalo de confiança |
| 04 — Diferenças-em-Diferenças | Setup 2×2; tendências paralelas; DiD via regressão; TWFE; erros-padrão clusterizados; teste placebo; problema do TWFE escalonado (Goodman-Bacon); Callaway & Sant'Anna; controle sintético | Método 2: hipótese de tendências paralelas examinada (pré-tendências ou estudo de eventos), TWFE com erros clusterizados, placebo e, se a adoção for escalonada, um estimador robusto à heterogeneidade |

## 2. Entrega 1 — o caderno: do download às análises

O caderno é a peça central do trabalho e deve conter **todas** as etapas, em sequência, em um único arquivo ou em poucos cadernos numerados (`01_...`, `02_...`). A regra é simples: se um número aparece no artigo, o código que o produziu está no caderno; se um arquivo é lido pelo caderno, o código que o baixou e preparou também está no caderno.

Etapas obrigatórias, na ordem em que devem aparecer:

1. **Identificação e pergunta.** Cabeçalho com disciplina, integrantes, pergunta de pesquisa, estimando (ATE, ATT etc.), DAG da relação causal com os controles escolhidos (Módulo 01) e quais métodos serão aplicados.
2. **Preparação do ambiente.** Instalação das bibliotecas (`pip install` dentro do caderno), importações, definição de caminhos relativos e de uma semente aleatória.
3. **Download dos dados brutos.** Código que baixa os arquivos diretamente das fontes originais (DATASUS, IBGE, MDS, portais de dados abertos, APIs). Nenhum arquivo de entrada pode chegar "pronto" por fora do caderno. Se a fonte exige interação manual (login, captcha), isso deve ser documentado e justificado, e o arquivo entregue junto, com sua origem e data de obtenção.
4. **Descompressão e seleção dos arquivos.** Extração de `.zip`, `.rar`, `.7z` etc., remoção do que não será usado e manutenção apenas dos arquivos necessários à análise, em pasta organizada (ver seção 3).
5. **Documentação das variáveis.** Exportação de um dicionário de colunas (nome, tipo, rótulo) e dos rótulos de valores das variáveis categóricas, antes de qualquer transformação.
6. **Construção e validação da base analítica.** Junções, agregações, criação de tratamento, desfecho e covariáveis, verificando dimensões, chaves únicas e ausência de valores faltantes inesperados.
7. **Análise descritiva.** Estatísticas e gráficos de tratamento, desfecho e covariáveis, com interpretação explícita de que ainda não há leitura causal.
8. **Métodos causais (Módulos 03 e 04).** Hipóteses de identificação enunciadas, diagnósticos (balanceamento, suporte comum, tendências paralelas, placebos), estimativas e intervalos de confiança.
9. **Robustez e poder.** Especificações alternativas e, quando cabível, cálculo do efeito mínimo detectável.
10. **Síntese e exportação.** Tabela consolidada de resultados e gravação das tabelas (`.csv`) e figuras (`.png`) usadas no artigo, com os mesmos nomes citados nele.

O caderno deve ser entregue **executado**, com todas as saídas visíveis (células sem saída ou com erro não serão consideradas). Antes de entregar, reinicie o kernel e execute tudo de uma vez ("Restart & Run All").

## 3. A etapa inicial em detalhe: download, descompressão e seleção dos dados

Essa é a etapa que os grupos mais tendem a deixar fora do caderno, e é justamente a que garante que outra pessoa consiga refazer o trabalho. O caderno `day_01.ipynb`, disponibilizado pelo professor, mostra o passo a passo com os microdados da Avaliação de Impacto do Bolsa Família (AIBF II, SAGI/MDS) e serve de modelo para esta parte. Ele faz, nesta ordem:

| Passo | O que o código faz | Por que importa |
| --- | --- | --- |
| 1. Download | Baixa o `.zip` da URL oficial com `requests`, em blocos (`stream=True`), e grava em disco | A origem do dado fica registrada no próprio código, com a URL exata |
| 2. Extração do zip | Extrai para uma pasta `raw/` e apaga o `.zip` | Libera espaço e evita arquivos duplicados |
| 3. Verificação de dependência externa | Testa se `unrar` está instalado e, se não, imprime instruções por sistema operacional | Quem reexecuta sabe o que precisa instalar antes |
| 4. Extração dos `.rar` internos | Percorre `raw/` com `os.walk`, extrai cada `.rar` e o remove | Fontes públicas frequentemente aninham compactações |
| 5. Extração seletiva | Abre apenas os `.zip` de interesse (`education.zip`, `descricao das variaveis.zip`) e apaga todos os demais | Só o necessário à análise permanece |
| 6. Lista de arquivos permitidos | Mantém somente os arquivos de uma lista explícita (`education.dta` e o `.log` com a descrição das variáveis); remove o resto | A decisão de quais dados usar fica explícita e auditável |
| 7. Organização | Move os arquivos mantidos para `raw/day 01/` e remove diretórios vazios | Estrutura de pastas previsível |
| 8. Leitura e documentação | Lê o `.dta` sem converter categóricas, exporta dicionário de colunas (`educacao_colunas.csv`), rótulos de valores (`educacao_value_labels.txt`), uma amostra e a base completa em `.csv` para `clean/day 01/` | Separa dado bruto (`raw/`) de dado tratado (`clean/`) e documenta os códigos antes de recodificar |

O que se espera do grupo nesta etapa:

- **URL ou chamada de API explícita no código** para cada fonte usada, com a data em que foi acessada em célula markdown.
- **Extração programática** de todos os arquivos compactados, com registro de dependências externas (`unrar`, `7z`) quando necessárias.
- **Seleção explícita** dos arquivos mantidos, em uma lista no código, e remoção do restante. Não deixe gigabytes de dados não usados na pasta do projeto.
- **Duas pastas separadas**: `raw/` (dados como baixados, nunca editados) e `clean/` ou `data/processed/` (bases produzidas pelo caderno).
- **Dicionário de variáveis e rótulos de valores exportados** antes da limpeza, para que o leitor saiba o que cada código significa.
- **Microdados pesados** (SIM, SINASC, CadÚnico, Censo): a agregação para a unidade de análise (município-ano, por exemplo) é feita no caderno, não fora dele. Se o processamento for longo, use um caderno `01_extracao.ipynb` separado, e grave os agregados em `data/processed/`; o caderno de análise os lê de lá. Os dois cadernos são entregues.

Quando a fonte oferecer uma biblioteca de acesso (por exemplo, `PySUS` para o DATASUS, `sidrapy` ou `ipeadatapy` para IBGE/Ipea, `basedosdados` para o BigQuery da Base dos Dados), seu uso é bem-vindo, desde que a chamada esteja no caderno e a versão do pacote esteja registrada.

## 4. Estrutura e boas práticas do caderno

O caderno é lido por uma pessoa, não só executado por uma máquina. Intercale células markdown e células de código: antes de cada bloco, diga o que será feito e por quê; depois, diga o que o resultado mostra. Um caderno só com código não atende ao esperado.

- **Estrutura de pastas fixa**, com o caderno em `notebooks/` e caminhos relativos à raiz do projeto (`Path("..") / "data" / "raw"`). Nunca use caminhos absolutos do seu computador (`C:/Users/...`, `/Users/nome/...`).
- **Instalação das dependências no próprio caderno** (`!pip install ...`) ou em um `requirements.txt` entregue junto, com versões fixadas.
- **Semente aleatória definida** (`SEMENTE = 42`) e usada em amostragens, bootstrap e pareamentos.
- **Validações com `assert`** após cada construção de base: número de linhas esperado, chave única, ausência de faltantes. Se algo falhar, o caderno para ali, em vez de produzir resultados sobre uma base quebrada.
- **Identificadores lidos como texto** (códigos IBGE de município, NIS, CNES), nunca como número, para não perder zeros à esquerda.
- **Decisões de dados registradas em markdown** no ponto em que são tomadas: exclusão de registros, imputações, proxies, definição de limiares. Cada uma com a justificativa e a contagem de registros afetados.
- **Funções para o que se repete** (estimar um modelo, registrar um resultado em tabela de síntese), em vez de copiar e colar células.
- **Saídas controladas**: evite imprimir dataframes inteiros; use `head()`, `describe()` e tabelas-resumo. Suprima avisos irrelevantes.
- **Figuras e tabelas salvas em arquivo** (`outputs/figuras/fig01_....png`, `outputs/tabelas/tab01_....csv`) com os mesmos números usados no artigo.
- **Seções numeradas** com títulos `##`, para que o artigo possa remeter ao caderno ("ver Seção 8 do caderno").

## 5. Entrega 2 — o relatório em formato de artigo científico

O relatório deve ter a forma de um **artigo científico curto**, entre 8 e 15 páginas (sem contar referências e apêndices), escrito para um leitor que não abrirá o caderno. Ele não é um relato do que o grupo fez ("primeiro baixamos, depois limpamos"), e sim a apresentação de uma pergunta, de um desenho de pesquisa e de uma resposta com suas incertezas.

| Seção | Conteúdo esperado |
| --- | --- |
| Título, autores, filiação | Título informativo (política, desfecho, lugar, período); nomes completos; disciplina e programa |
| Resumo (até 250 palavras) e palavras-chave | Pergunta, dados, métodos, resultado principal com magnitude e intervalo, conclusão |
| 1. Introdução | A política e seu mecanismo esperado; o que a literatura já encontrou (ao menos 3 referências); a pergunta causal; o estimando (ATE, ATT, ITT); a unidade de análise |
| 2. Dados | Fontes (com link e data de acesso), período, cobertura; construção de tratamento, desfecho e covariáveis; decisões de limpeza e seus efeitos na amostra; estatísticas descritivas |
| 3. Estratégia empírica | Para cada método: hipóteses de identificação em linguagem clara, equação estimada, erros-padrão, diagnósticos planejados (balanceamento, suporte comum, pré-tendências, placebos) |
| 4. Resultados | Tabelas com estimativa, erro-padrão e IC 95%; figuras de diagnostico (love plot, estudo de eventos). |
| 5. Robustez | Especificações alternativas, definições alternativas de tratamento, poder estatístico ou efeito mínimo detectável |
| 6. Discussão e limitações | O que os resultados permitem e não permitem afirmar; ameaças à identificação; problemas de medição; o que seria preciso para avançar |
| 7. Conclusão | Resposta direta à pergunta, em um ou dois parágrafos, com implicações para a política |
| Declaração de uso de IA | Obrigatória (seção 6 deste documento) |
| Referências | Normas ABNT ou APA, aplicadas de forma consistente; inclua os artigos metodológicos dos estimadores usados |
| Apêndice (opcional) | Tabelas adicionais, dicionário de variáveis, link para o repositório |

Regras de forma:

- Toda tabela e figura tem número, título, fonte e nota; é citada no texto; e corresponde a um arquivo gerado pelo caderno. Os números do artigo devem ser idênticos aos do caderno.
- Resultados não significativos são reportados com a mesma clareza dos significativos. Ausência de significância não é evidência de ausência de efeito, e o artigo deve discutir poder.
- A linguagem causal ("efeito", "impacto", "reduziu") só aparece onde o desenho a sustenta; a descritiva usa "associação", "correlação", "evolução".
- Escrita impessoal ou na primeira pessoa do plural, sem relatos de bastidores ("tivemos dificuldade com o unrar"). Esses relatos ficam no caderno, em markdown.
- Formato: `.docx` ou `.pdf`, fonte 11 ou 12, espaço 1,5, margens de 2,5 cm.

## 6. Uso de inteligência artificial generativa

O trabalho segue a **Política de Uso de Inteligência Artificial na Pós-Graduação Stricto Sensu do IDP** (versão 1.0, 2026), que se aplica ao Mestrado e Doutorado em Administração (art. 2º, VI). As regras abaixo são as da Política, com as especificações que o art. 12, IV, autoriza o docente a fazer para a disciplina. Leiam o documento na íntegra; em caso de dúvida sobre um uso, presume-se a necessidade de declaração (art. 5º, § 2º).

O princípio central é o da **autoria e responsabilidade humana** (art. 4º, I): a IA não é autora, e o grupo responde pela exatidão de tudo o que entregar, inclusive do que foi gerado ou revisado com auxílio de IA. Em arguição, qualquer integrante deve ser capaz de explicar qualquer linha do caderno e qualquer frase do artigo.

**Como a Política classifica os usos (arts. 6º a 8º), aplicada a este trabalho:**

| Categoria | Exemplos neste trabalho | O que é exigido |
| --- | --- | --- |
| Incentivados (art. 6º) | Discutir a pergunta e o desenho com a IA como "oponente socrático"; mapear literatura; gerar e depurar código de download, limpeza, gráficos e análise exploratória; pedir sugestões de correção linguística no diálogo, sem alterar o arquivo | Verificação crítica: toda referência conferida na fonte original; todo código lido, testado e entendido antes de entrar no caderno. Declaração quando o uso for substantivo (art. 3º, VI) |
| Condicionados (art. 7º) | Revisão linguística com alteração direta do texto; redação de trechos como apoio; interpretação e análise de dados com IA; geração de tabelas ou figuras | Declaração obrigatória; reescrita crítica pelo grupo; método compreendido e controlado pelos autores; transparência sobre prompts e saídas relevantes; figuras e tabelas geradas por IA identificadas como tal e justificadas |
| Vedados (art. 8º) | Delegar à IA partes essenciais do trabalho (pergunta, desenho, interpretação, conclusão); apresentar conteúdo gerado como próprio sem revisão e declaração; fabricar ou manipular dados, resultados, citações ou referências; listar a IA como autora; carregar dados pessoais, sensíveis ou não publicados em ferramentas sem garantias | Caracteriza má conduta acadêmica, apurada conforme o art. 16 e as normas de integridade do IDP |

**Dados.** Os microdados usados na disciplina são públicos, mas a regra do art. 4º, VI, vale igualmente: não carreguem em ferramentas de terceiros bases com identificação pessoal (CPF, NIS, nome, endereço) nem dados obtidos por acesso restrito, salvo garantia contratual de não retenção e de não uso para treinamento. Para pedir ajuda com código, use amostras anonimizadas ou apenas a estrutura das colunas, como o dicionário de variáveis gerado na etapa inicial.

**Declaração de uso de IA (arts. 9º e 10).** Todo grupo que fizer uso substantivo de IA inclui uma declaração em dois lugares: no cabeçalho do caderno e, no artigo, como seção própria logo antes das referências, conforme o modelo do Anexo II da Política. A declaração contém, no mínimo:

1. a ferramenta, o modelo e a versão utilizados, com a data de uso;
2. o propósito de cada uso (exploração de ideias, busca de literatura, geração de código, revisão linguística, apoio à análise);
3. a forma de emprego e as etapas do trabalho em que a ferramenta foi usada (download, limpeza, estimação, redação);
4. a afirmação de que os autores revisaram e editaram criticamente o conteúdo e assumem a responsabilidade pelo trabalho.

A não utilização também deve ser declarada (art. 10, § 3º). Guardem os prompts e as saídas relevantes das interações: o professor pode solicitá-los como material complementar para verificar transparência e replicabilidade (art. 10, § 2º).

**Especificações da disciplina.** Além do que a Política já exige:

- A **reexecução integral do caderno pelos autores** é parte da revisão crítica. Código gerado por IA que o grupo não sabe explicar é tratado como conteúdo não revisado (art. 8º, II).
- A **interpretação dos resultados e a conclusão** são dos autores. A IA pode ser consultada sobre elas, mas o texto final deve refletir o julgamento do grupo sobre o que os diagnósticos e as estimativas permitem afirmar.
- **Referências bibliográficas** são verificadas uma a uma na fonte original. Referência inexistente ou que não diz o que se afirma é hipótese vedada pelo art. 8º, VI, e será encaminhada conforme o art. 16.
- Ferramentas de detecção de IA podem ser usadas pelo professor com caráter indiciário, nunca conclusivo (art. 16, § 1º); a análise é sempre humana e contextual, e o grupo terá oportunidade de esclarecimento.

## 7. Formato de entrega e checklist final

Entregue uma única pasta compactada (`.zip`) ou o link de um repositório público (GitHub/GitLab), nomeada `TrabalhoFinal_APPD_<sobrenomes>`, com a estrutura abaixo. Os dados brutos (`data/raw/`) **não** são entregues: o caderno os baixa.

```markdown
TrabalhoFinal_APPD_<sobrenomes>/
├── README.md                  # pergunta, integrantes, como executar, tempo estimado, dependências externas
├── requirements.txt           # pacotes com versões
├── notebooks/
│   ├── 01_extracao.ipynb      # download, descompressão, seleção, agregação (executado)
│   └── 02_analise.ipynb       # painel, descritiva, métodos, robustez, síntese (executado)
├── data/
│   ├── raw/                   # vazio na entrega; preenchido pelo caderno 01
│   └── processed/             # bases geradas pelos cadernos
├── outputs/
│   ├── figuras/               # fig01_..., fig02_... usadas no artigo
│   └── tabelas/               # tab01_..., tab02_... usadas no artigo
└── artigo/
    └── Artigo_<sobrenomes>.pdf
```

Um único caderno é aceito se o processamento for leve; a divisão em dois é recomendada quando os microdados forem pesados.

Antes de entregar, confira:

- [ ] Arquivos compactados são extraídos por código e apenas os necessários são mantidos
- [ ] Dicionário de variáveis e rótulos de valores foram exportados
- [ ] Caminhos são relativos; não há caminho do computador de ninguém
- [ ] "Restart & Run All" executa do início ao fim sem erro, e as saídas estão salvas
- [ ] Há `assert` validando a base analítica
- [ ] Ao menos dois métodos da disciplina, com hipóteses enunciadas e diagnósticos
