```python
import pandas as pd
```

```python
import os
import requests
import shutil
```

```python
url = "https://raw.githubusercontent.com/vaibhavwalvekar/Econometric-Analysis-Progresa-Mexico/master/progresa_sample.csv"
csv_path = os.path.join("raw", "day 02", "progresa_sample.csv")
extract_dir = "raw/day 02"
os.makedirs(extract_dir, exist_ok=True)
```

```python
# ==============================
# 1. baixar dados
# ==============================

print("Baixando arquivo...")
with requests.get(url, stream=True) as r:
    r.raise_for_status()
    with open(csv_path, "wb") as f:
        for chunk in r.iter_content(chunk_size=8192):
            f.write(chunk)

print("Download concluído.")
```

```python
# ==============================
# 3. carregar o CSV
# ==============================

df = pd.read_csv("raw/day 02/progresa_sample.csv")
```

```python
os.makedirs("clean/day 02/", exist_ok=True)

df_colunas = pd.DataFrame({
    "coluna": df.columns,
    "dtype": df.dtypes.astype(str),
})

df_colunas.to_csv("clean/day 02/progresa_colunas.csv", index=False, encoding="utf-8")
```

```python
# (2) Amostra com todas as colunas

df_amostra = df.sample(min(3000, len(df)), random_state=42)
df_amostra.to_csv("clean/day 02/progresa_amostra.csv", index=False, encoding="utf-8")
```

```python
df.to_csv("clean/day 02/progresa_completo.csv", index=False, encoding="utf-8")
```
